```python
from pprint import pprint

daily_records =[{"data":"2026-09-25","food":80,"traffic":20,"shopping":0},
        {"data":"2026-09-26","food":120,"traffic":2,"shopping":110},
        {"data":"2026-09-27","food":10,"traffic":222,"shopping":10},
]
print(f"show me a everythreeday\n")
pprint(f"{daily_records}\n")

#1计算所有天的总支出
#daily_records[0]，测试一下如何正确访问列表嵌套字典元素的单个项
print(f"{daily_records[0]["data"]}")
#用累加器total_expenses循环累加最后得出这三天的总支出，只用=号是错误的数据
#到最后再考虑其他方式实现
total_expenses = 0
for records in daily_records:
  total_expenses += records["food"]+records["traffic"]+records["shopping"]
print(total_expenses)

#2找出哪一天花最多，花多少。金额最大最高且金额是多少
# max_expense = max(daily_records,key=daily_records.get)
# print(max_expense)
#虽然我的任务是要找到哪天花销最大，但是我也不能一上来就去找到最大的花销吧
#因此我要首先把每天的总花销计算出来，再进行内循环内值得比较，才会显地更合理，有步骤也会更容易实现

max_expense = -1
maxexpense_data = ""
for record in daily_records:
  day_totalexpense = record["food"]+record["traffic"]+record["shopping"]
  if day_totalexpense > max_expense:
    max_expense = day_totalexpense
    maxexpense_data = record["data"]
print(f"花销最大的一天是{maxexpense_data},且花销总金额是{day_totalexpense:.2f}元")

```
